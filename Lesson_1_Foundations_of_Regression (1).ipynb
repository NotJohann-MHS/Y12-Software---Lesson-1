## Part 2: Visualising Data with Matplotlib

Before training a machine learning model, we can visualise the relationship between variables to identify patterns, trends and unusual values. 

Linear regression is particularly suitable when the relationship between the variables is approximately linear.

```python
import matplotlib.pyplot as plt

hours = np.array([1, 2, 3, 4, 5, 0])
scores = np.array([50, 55, 100, 70, 80, 30])

plt.scatter(hours, scores)
plt.xlabel('Hours Studied')
plt.ylabel('Test Score')
plt.title('Study Time vs Test Score')
plt.show()
```

#### Extension Dataset
Exercise hours: 1,2,3,4,5
Calories burned: 200,250,300,360,420
- Create a scatter plot

```python
E_Hours = np.array([1,2,3,4,5])
Calories = np.array([200,250,300,360,420])
plt.scatter(Calories, E_Hours)
plt.ylabel('Calories burned')
plt.xlabel('Hours exercised')
plt.title('Hours exercised vs Calories burned')
plt.show()


from sklearn.linear_model import LinearRegression

# Prepare the input data
X = E_Hours.reshape(-1, 1)
y = Calories

# Create the model
model = LinearRegression()

# Train the model using the known data
model.fit(X, y)

# Make a prediction - use the trained model to predict the test score for 6 hours study
prediction = model.predict([[6]])

print('Predicted calories burned for 6 hours:', prediction)

plt.scatter(X, y)
plt.plot(X, model.predict(X))
plt.ylabel('Calories burned')
plt.xlabel('Hours exercised')
plt.title('Linear Regression Model')
plt.show()
print("Slope:", model.coef_[0])
print("Intercept:", model.intercept_)
```

## Part 3: Linear Regression with Scikit-Learn

Term	             Meaning in this example
Feature (X)	         The input used to make a prediction — hours studied
Target (y)	         The actual value the model learns to predict — test score
Training	         Using known hours and scores to learn the relationship
Model	             The mathematical relationship learned from the data
Prediction	         The estimated score for a new number of study hours

Linear regression finds a straight line that best fits the training data by minimising the sum of squared residuals.

Linear regression attempts to model the relationship:
y = mx + b

This is supervised learning:
- X = input
- y = output


```python
from sklearn.linear_model import LinearRegression

# Prepare the input data
X = hours.reshape(-1, 1)
y = scores

# Create the model
model = LinearRegression()

# Train the model using the known data
model.fit(X, y)

# Make a prediction - use the trained model to predict the test score for 6 hours study
prediction = model.predict([[6]])

print('Predicted score for 6 hours:', prediction)
```

```python
plt.scatter(X, y)
plt.plot(X, model.predict(X))
plt.xlabel('Hours Studied')
plt.ylabel('Test Score')
plt.title('Linear Regression Model')
plt.show()
print("Slope:", model.coef_[0])
print("Intercept:", model.intercept_)
```
